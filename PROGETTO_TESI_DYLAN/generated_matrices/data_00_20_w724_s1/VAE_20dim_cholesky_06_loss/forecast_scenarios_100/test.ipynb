```python
from pathlib import Path
import torch

train_path = Path.cwd() / "forecast_scenarios_100.pt"

try:
    payload = torch.load(train_path, map_location="cpu", weights_only=False)
except TypeError:
    payload = torch.load(train_path, map_location="cpu")

print(f"File: {train_path}")
print(f"Type: {type(payload)}")

if isinstance(payload, dict):
    print(f"Keys: {list(payload.keys())}")
    for key, value in payload.items():
        if hasattr(value, "shape"):
            print(f"{key}: {type(value)} shape={tuple(value.shape)}")
        elif isinstance(value, (list, tuple)):
            print(f"{key}: {type(value)} len={len(value)}")
        else:
            print(f"{key}: {type(value)} value={value}")
elif isinstance(payload, torch.Tensor):
    print(f"Shape: {tuple(payload.shape)}")
    print(payload[:1] if payload.ndim > 0 else payload)
else:
    print(payload)

```
