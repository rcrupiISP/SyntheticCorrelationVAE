# Data Extraction from Compressed CSV

This notebook extracts data from the compressed `data.csv.gz` file and saves the processed data to the `data/processed` directory.

## 1. Import Required Libraries

```python
import pandas as pd
import os
from pathlib import Path
```

## 2. Load Compressed CSV File

Pandas automatically handles gzip compression when reading files with the `.gz` extension.

```python
# Define the path to the compressed CSV file
FILE_NAME = 'data_13_17'
file_path = f'../data/raw/{FILE_NAME}.csv.gz'

# Read the compressed CSV file
df = pd.read_csv(file_path)

print(f"Data loaded successfully!")
print(f"Total rows: {len(df)}")
print(f"Total columns: {len(df.columns)}")
```

## 3. Explore Data Structure

Let's examine the first rows and get information about the dataset.

```python
# Display first few rows
print("First 5 rows:")
print(df.head())
```

```python
# Get data info
print("\nDataset Info:")
df.info()
```

```python
# Get statistical summary
print("\nStatistical Summary:")
print(df.describe())
```

## 4. Save Processed Data

Create the output directory if it doesn't exist and save the extracted data as CSV.

```python
# Define the output directory
output_dir = Path('../data/processed')

# Create the directory if it doesn't exist
output_dir.mkdir(parents=True, exist_ok=True)

# Save the data as CSV
output_path = output_dir / f'{FILE_NAME}.csv'
df.to_csv(output_path, index=False)

print(f"Data saved to: {output_path}")
print(f"File size: {os.path.getsize(output_path) / (1024*1024):.2f} MB")
```
